```python
from pylatex import Command, Document, Section, Subsection, Package
from pylatex.base_classes import Environment
from pylatex.beamer import Frame

from pylatex.utils import NoEscape, italic
```

```python
class AndurilPresentation(Document):
    def __init__(self, document_options=None):
        super().__init__(documentclass="beamer", inputenc=None, document_options=document_options,
                         page_numbers=True)
        self.preamble.append(Command('usetheme','anduril'))
        self.preamble.append(Command('usefonttheme','professionalfonts'))
        self.preamble.append(Command('usefonttheme','structurebold'))

        self.preamble.append(Package('fontspec'))
        self.preamble.append(Command('setmainfont', 'Inter'))
```

```python
doc = AndurilPresentation(document_options='14pt,aspectratio=1609')

doc.preamble.append(Command('title',"Awesome Anduril Presentation"))
doc.preamble.append(Command("author", "Anonymous author"))
doc.preamble.append(Command("date", NoEscape(r"\today")))

#doc.append(Command('setbeamercolor', 'background canvas', extra_arguments='bg=black'))
#doc.append(Command('setbeamercolor', 'title', extra_arguments='fg=white'))
#doc.append(Command('setbeamercolor', 'subtitle', extra_arguments='fg=white'))
#doc.append(Command('setbeamercolor', 'author', extra_arguments='fg=white'))
#doc.append(Command('setbeamercolor', 'date', extra_arguments='fg=white'))
with doc.create(Frame()):
    doc.append(NoEscape(r"\titlepage"))
#doc.append(Command('setbeamercolor', 'background canvas', extra_arguments='bg=white'))

with doc.create(Frame('Outline')):
    doc.append(NoEscape(r"\tableofcontents"))


doc.append(Section("Section 1"))
with doc.create(Frame(title='Slide 1')):
    doc.append("Some content")

doc.append(Section("Section 2"))
with doc.create(Frame(title='Slide 2.1')):
    doc.append("Some content")

doc.append(Section("Section 3"))
with doc.create(Frame(title='Slide 3.1')):
    doc.append("Some content")


doc.generate_pdf('./build/test_beamer', clean_tex=False,
                 compiler=(("latexmk", []), ("lualatex", []))
                 )
```
